## 🌸 Dataset

The **Iris Dataset** is a popular Machine Learning dataset containing measurements of iris flowers from three different species.

### Features

- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

### Target

- Species

The dataset will be used to understand different types of visualizations and relationships between numerical features.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target

df.head()
```

## 🔍 Dataset Overview

Before visualization, it is important to understand the structure of the dataset.

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

# 📈 1. Line Chart

A line chart is generally used to visualize trends or changes across an ordered sequence.

Here, we will visualize the Sepal Length values for the first 30 observations.

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df.index[:30],
    df["sepal length (cm)"][:30],
    marker="o"
)

plt.title("Sepal Length - First 30 Observations")
plt.xlabel("Observation")
plt.ylabel("Sepal Length (cm)")

plt.grid(True)
plt.show()
```

# 📊 2. Histogram

A histogram shows the distribution of numerical data.

It helps us understand:

- Data distribution
- Frequency
- Range
- Concentration of values

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["sepal length (cm)"],
    bins=10
)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.show()
```

# 🔵 3. Scatter Plot

A scatter plot is used to visualize the relationship between two numerical variables.

Here, we compare:

**Sepal Length vs Petal Length**

This helps us visually identify whether the two features have a relationship.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="species"
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")

plt.show()
```

# 🔥 4. Correlation Heatmap

Correlation measures the relationship between numerical variables.

A correlation value can range from:

- **+1** → Strong positive correlation
- **0** → No linear correlation
- **-1** → Strong negative correlation

A heatmap makes these relationships easier to understand visually.

```python
correlation = df.drop(columns=["species"]).corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```
